```python
from COMP import *
```

```python
hw = py4hw.HWSystem()

i1 = hw.wire('i1', 32)
i2 = hw.wire('i2', 32)
out = hw.wire('out', 32)
unsigned_lt = hw.wire('unsigned_lt', 32)
signed_lt = hw.wire('signed_lt', 32)
lt = hw.wire('lt')
gt = hw.wire('gt')
eq = hw.wire('eq')

dut = COMP(hw, 'cmp', i1, i2, unsigned_lt, signed_lt, lt, gt, eq)
```

```python
i1.put(23)
i2.put(-54)
hw.getSimulator().clk()
print('unsigned_lt', unsigned_lt.get(), 'signed_lt', signed_lt.get(), 'lt', lt.get(), 'gt', gt.get(), 'eq', eq.get()) 
```

```python
rtl = py4hw.VerilogGenerator(dut)
verilog = rtl.getVerilogForHierarchy(dut)

print(verilog)
```
