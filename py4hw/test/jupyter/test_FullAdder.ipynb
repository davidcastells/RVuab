```python
from FullAdder import *
```

```python
hw = py4hw.HWSystem()

reset = hw.wire('reset')
sign_extend = hw.wire('sign_extend')
i1 = hw.wire('i1')
i2 = hw.wire('i2')
ci = hw.wire('ci')
inv_i2 = hw.wire('inv_i2')
r = hw.wire('r')
co = hw.wire('co')
r_and = hw.wire('r_and')
r_xor = hw.wire('r_xor')

dut = FullAdder(hw, 'fa', i1, i2, ci, inv_i2, r, co, r_and, r_xor)
```

```python
for vinv_i2 in [0, 1]:
    for vi1 in [0, 1]:
        for vi2 in [0, 1]:
            for vci in [0, 1]:
                i1.put(vi1)
                i2.put(vi2)
                ci.put(vci)
                inv_i2.put(vinv_i2)

                hw.getSimulator().clk()

                vr = r.get()
                vco = co.get()
                vr_and = r_and.get()
                vr_xor = r_xor.get()

                exp_vr = ((vi1 + (vi2 ^ vinv_i2) + vci) & 1)
                exp_vco = (((vi1 + (vi2 ^ vinv_i2) + vci) ) > 1) & 1
                exp_vr_and = (vi1 & vi2)
                exp_vr_xor = (vi1 ^ vi2)
                ok = (vr ==  exp_vr)
                ok = ok and (vco == exp_vco)
                ok = ok and (vr_and == exp_vr_and)
                ok = ok and (vr_xor == exp_vr_xor)
                
                sok = '[OK]' if ok else '[ERROR]'
                print(vinv_i2, vi1, vi2, vci,  '|', vr, vco, vr_and, vr_xor, sok)

                
```

```python
rtl = py4hw.VerilogGenerator(dut)
verilog = rtl.getVerilogForHierarchy(dut)

print(verilog)
```
