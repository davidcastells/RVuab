```python
from SHIFT import *
```

```python
hw = py4hw.HWSystem()

reset = hw.wire('reset')
sign_extend = hw.wire('sign_extend')
i1 = hw.wire('i1', 32)
i2 = hw.wire('i2', 32)
sr = hw.wire('sr', 32)
sra = hw.wire('sra', 32)
sl = hw.wire('sl', 32)

SHIFT(hw, 'shift', i1, i2, sign_extend, sr, sra, sl)
```

```python
i1.put(0xAABBCC)
i2.put(8)
hw.getSimulator().clk()

vsr, vsra, vsl = sr.get(), sra.get(), sl.get()

print(hex(vsr), hex(vsra), hex(vsl))
```

```python
i1.put(0xFFAABBCC)
i2.put(8)
hw.getSimulator().clk()

vsr, vsra, vsl = sr.get(), sra.get(), sl.get()

print(hex(vsr), hex(vsra), hex(vsl))

```
