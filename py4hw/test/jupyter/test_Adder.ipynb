```python
from Adder import *
```

```python
hw = py4hw.HWSystem()

i1 = hw.wire('i1', 32)
i2 = hw.wire('i2', 32)
out = hw.wire('out', 32)

dut = Adder(hw, 'adder', i1, i2, out)
```

```python
i1.put(23)
i2.put(54)
hw.getSimulator().clk()
assert(out.get() == 77)
```

```python
i1.put(23)
i2.put(-24)
hw.getSimulator().clk()
assert(out.get() == 0xffffffff )
```

```python
rtl = py4hw.VerilogGenerator(dut)
verilog = rtl.getVerilogForHierarchy(dut)

print(verilog)
```
