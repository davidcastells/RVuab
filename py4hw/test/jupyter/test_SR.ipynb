```python
from SR import *
```

```python
hw = py4hw.HWSystem()

reset = hw.wire('reset')
sign_extend = hw.wire('sign_extend')
i1 = hw.wire('i1', 32)
i2 = hw.wire('i2', 32)
inv_i2 = hw.wire('inv_i2') # invert i2 (subtraction) and use as initial carry-in
zero_i20 = hw.wire('zero_i20') # force LSB of result to 0
r = hw.wire('r', 32)
r_and = hw.wire('r_and', 32)
r_xor = hw.wire('r_xor', 32)

 
dut = SR(hw, 'sr', i1, i2, inv_i2, zero_i20, r, r_and, r_xor)
```

```python
i1.put(0x111)
i2.put(0x222)
inv_i2.put(0)
zero_i20.put(0)

hw.getSimulator().clk()

print(f'i1:{i1.get():08X} i2:{i2.get():08X} inv_i2:{inv_i2.get():X} zero_i20:{zero_i20.get():X} r:{r.get():08X} r_and:{r_and.get():08X} r_xor:{r_xor.get():08X}')
```

```python
i1.put(0x111)
i2.put(0x222)
inv_i2.put(1)
zero_i20.put(0)

hw.getSimulator().clk()

print(f'i1:{i1.get():08X} i2:{i2.get():08X} inv_i2:{inv_i2.get():X} zero_i20:{zero_i20.get():X} r:{r.get():08X} r_and:{r_and.get():08X} r_xor:{r_xor.get():08X}')
```
