```python
from RegFile import *
```

```python
hw = py4hw.HWSystem()

reset = hw.wire('reset')
rs1_addr = hw.wire('rs1_addr', 5)
rs2_addr = hw.wire('rs2_addr', 5)
rd_addr = hw.wire('rd_addr', 5)
ld_rd = hw.wire('ld_rd')
rd = hw.wire('rd', 32)
rs1 = hw.wire('rs1', 32)
rs2 = hw.wire('rs2', 32)

RegFile(hw, 'regfile',   reset, rs1_addr, rs2_addr, rd_addr, ld_rd, rd, rs1, rs2)
```

```python
# Write values
for i in range(32):
    rd_addr.put(i)
    ld_rd.put(1)
    rd.put(0x10000 + i)
    hw.getSimulator().clk()
    ld_rd.put(0)
    hw.getSimulator().clk()

# Check written values in rs1
for i in range(32):
    rs1_addr.put(i)
    rs2_addr.put(31-i)
    hw.getSimulator().clk()
    print(f'rs1[{i}] = {rs1.get():08X}\trs2[{31-i}] = {rs2.get():08X}')
    if (i == 0):
        assert(rs1.get() == 0)
    else:
        assert(rs1.get() == 0x10000 + i)

    if (31-i == 0):
        assert(rs2.get() == 0)
    else:
        assert(rs2.get() == 0x10000 + (31-i))
```
